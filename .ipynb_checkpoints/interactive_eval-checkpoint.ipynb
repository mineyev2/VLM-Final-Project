# Interactive Evaluation Notebook

This notebook allows you to **load the model once** and keep it in GPU memory while running evaluations multiple times without reloading.

**Key Benefits:**
- Model stays in GPU memory across cells
- Fast iteration on different checkpoints
- Quick experimentation with generation parameters
- Inline visualizations

## 1. Setup and Imports

```python
import os
import re
import torch
import torch.nn as nn
import numpy as np
import cv2
import matplotlib.pyplot as plt
from PIL import Image
from tqdm import tqdm
from termcolor import colored

from src.models.qwen_clip_model import QwenCLIPModel
from src.utils.utils import ProjectWorldToImage, OffsetTrajectory3D
from nuscenes import NuScenes
from pyquaternion import Quaternion

# Set device
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Using device: {device}")
```

## 2. Load Model (ONCE - Stays in GPU Memory)

**Run this cell once** at the start. The model will stay in memory for all subsequent cells.

```python
# Initialize model (loads base weights - this takes time but only happens once!)
qwen_model_name = "Qwen/Qwen3-4B"  # or "Qwen/Qwen2.5-3B-Instruct"
model = QwenCLIPModel(device, qwen_model_name=qwen_model_name)
model.eval()

print(colored("✓ Model loaded and in GPU memory!", "green"))
```

## 3. Load Checkpoint

Load trained weights into the model (fast - just updates weights, not reloading model).

```python
# Load checkpoint weights
checkpoint_path = "./experiments/runs/qwenclip2a/final_model.pth"

data = torch.load(checkpoint_path, map_location=device)
if 'language_model_state_dict' in data:
    model.language_model.load_state_dict(data['language_model_state_dict'], strict=False)
    model.vision_tower.load_state_dict(data['vision_tower_state_dict'])
    model.mlp_projector.load_state_dict(data['model_state_dict'])
    print(colored("✓ Loaded full checkpoint (projector + vision tower)", "green"))
else:
    model.mlp_projector.load_state_dict(data)
    print(colored("✓ Loaded projector checkpoint", "green"))
```

## 4. Initialize NuScenes Dataset

```python
dataroot = '/storage/ice-shared/cs8803vlm/rmineyev3'
version = 'v1.0-test'

nusc = NuScenes(version=version, dataroot=dataroot, verbose=False)

# Build sample list (same as training)
sample_tokens = []
for scene in nusc.scene:
    nbr_samples = scene['nbr_samples']
    if nbr_samples < 13:
        continue
    first_sample_token = scene['first_sample_token']
    sample = nusc.get('sample', first_sample_token)
    for _ in range(2):
        sample = nusc.get('sample', sample['next'])
    for _ in range(nbr_samples - 12):
        sample_tokens.append(sample['token'])
        sample = nusc.get('sample', sample['next'])

print(f"✓ Dataset initialized with {len(sample_tokens)} samples")
```

## 5. Helper Functions

```python
def parse_coords_from_text(text, max_points=10):
    """Extract waypoint coordinates from generated text."""
    trajectory_match = re.search(r'Future Trajectory:\s*(.+)', text, re.IGNORECASE | re.DOTALL)
    text_to_parse = trajectory_match.group(1) if trajectory_match else text
    
    nums = re.findall(r"[-+]?[0-9]*\.?[0-9]+(?:[eE][-+]?[0-9]+)?", text_to_parse)
    nums = [float(x) for x in nums]
    
    pairs = []
    for i in range(0, len(nums) - 1, 2):
        pairs.append([nums[i], nums[i+1]])
        if len(pairs) >= max_points:
            break
    
    return np.array(pairs, dtype=float) if len(pairs) > 0 else np.array([], dtype=float).reshape(0, 2)


def get_sample_data(idx):
    """Get image, ego positions, and waypoints for a sample."""
    sample_token = sample_tokens[idx]
    sample = nusc.get('sample', sample_token)
    
    # Ego positions (last 3)
    ego_positions = []
    current_sample = sample
    for _ in range(3):
        cam_data = nusc.get('sample_data', current_sample['data']['CAM_FRONT'])
        ego_pose = nusc.get('ego_pose', cam_data['ego_pose_token'])
        ego_positions.append([float(ego_pose['translation'][0]), float(ego_pose['translation'][1])])
        if current_sample['prev']:
            current_sample = nusc.get('sample', current_sample['prev'])
        else:
            while len(ego_positions) < 3:
                ego_positions.append(ego_positions[-1])
    ego_positions.reverse()
    
    # Image and calibration
    camera_token = sample['data']['CAM_FRONT']
    camera_data = nusc.get('sample_data', camera_token)
    image_path = os.path.join(nusc.dataroot, camera_data['filename'])
    image = Image.open(image_path).convert('RGB')
    
    cam_calib = nusc.get('calibrated_sensor', camera_data['calibrated_sensor_token'])
    ego_pose = nusc.get('ego_pose', camera_data['ego_pose_token'])
    cam_to_ego = {
        'translation': cam_calib['translation'],
        'rotation': cam_calib['rotation'],
        'camera_intrinsic': np.array(cam_calib['camera_intrinsic'])
    }
    ego_to_world = {
        'translation': ego_pose['translation'],
        'rotation': ego_pose['rotation']
    }
    
    # Ground truth waypoints (next 10)
    waypoints = []
    current_sample = sample
    for _ in range(10):
        next_sample = nusc.get('sample', current_sample['next'])
        next_camera_data = nusc.get('sample_data', next_sample['data']['CAM_FRONT'])
        next_ego_pose = nusc.get('ego_pose', next_camera_data['ego_pose_token'])['translation']
        waypoints.append([float(next_ego_pose[0]), float(next_ego_pose[1])])
        current_sample = next_sample
    
    return image, ego_positions, np.array(waypoints), cam_to_ego, ego_to_world


print("✓ Helper functions defined")
```

## 6. Test on a Single Sample

Run this cell multiple times with different `sample_idx` values to inspect outputs.

```python
# Select a sample
sample_idx = 0

# Get data
image, ego_positions, gt_waypoints, cam_to_ego, ego_to_world = get_sample_data(sample_idx)

# Prepare image tensor
pixel_values = model.image_processor(images=[image], return_tensors='pt').pixel_values.to(device)

# Generate trajectory
with torch.no_grad():
    outputs, gen_texts = model.generate_trajectory(pixel_values, [ego_positions])

gen_text = gen_texts[0]
pred_coords = parse_coords_from_text(gen_text, max_points=10)

# Print results
print(colored(f"Sample {sample_idx}:", "cyan", attrs=['bold']))
print(colored("Generated Text:", "yellow"))
print(gen_text[:500])  # First 500 chars
print(colored(f"\nParsed waypoints: {pred_coords.shape[0]}/10", "green"))
print(f"Ground truth shape: {gt_waypoints.shape}")

# Compute metrics
if pred_coords.shape[0] > 0:
    if pred_coords.shape[0] < 10:
        pad = np.full((10 - pred_coords.shape[0], 2), np.nan)
        pred_coords_padded = np.vstack([pred_coords, pad])
    else:
        pred_coords_padded = pred_coords[:10]
    
    diffs = pred_coords_padded - gt_waypoints
    l2_errors = np.linalg.norm(diffs, axis=1)
    ade = np.nanmean(l2_errors)
    fde = l2_errors[-1]
    
    print(colored(f"\nMetrics:", "magenta"))
    print(f"  ADE: {ade:.2f}m")
    print(f"  FDE: {fde:.2f}m")
```

## 7. Visualize Trajectory on Image

```python
# Convert to OpenCV format
img = cv2.cvtColor(np.array(image), cv2.COLOR_RGB2BGR)

# Convert 2D to 3D (add z=0)
gt_waypoints_3d = np.hstack([gt_waypoints, np.zeros((len(gt_waypoints), 1))])
pred_waypoints_3d = np.hstack([pred_coords, np.zeros((len(pred_coords), 1))])

# Filter out NaN predictions
valid_pred_mask = ~np.isnan(pred_waypoints_3d).any(axis=1)
pred_waypoints_3d_valid = pred_waypoints_3d[valid_pred_mask]

try:
    # Project GT to image
    gt_points_img = ProjectWorldToImage(gt_waypoints_3d.tolist(), cam_to_ego, ego_to_world)
    
    # Draw GT polygon (green)
    if len(gt_waypoints_3d) > 1:
        gt_left_3d = OffsetTrajectory3D(gt_waypoints_3d, -1.73 / 2)
        gt_right_3d = OffsetTrajectory3D(gt_waypoints_3d, 1.73 / 2)
        gt_left_img = ProjectWorldToImage(gt_left_3d.tolist(), cam_to_ego, ego_to_world)
        gt_right_img = ProjectWorldToImage(gt_right_3d.tolist(), cam_to_ego, ego_to_world)
        
        gt_polygon = np.vstack((np.array(gt_left_img), np.array(gt_right_img)[::-1])).astype(np.int32)
        frame_gt = np.zeros_like(img)
        cv2.fillPoly(frame_gt, [gt_polygon], color=(0, 255, 0))
        mask_gt = frame_gt.astype(bool)
        img[mask_gt] = cv2.addWeighted(img, 0.5, frame_gt, 0.5, 0)[mask_gt]
    
    # Draw GT waypoints
    for pt in gt_points_img:
        cv2.circle(img, tuple(pt.astype(int)), radius=8, color=(0, 255, 0), thickness=-1)
    
    # Draw predicted trajectory (orange)
    if len(pred_waypoints_3d_valid) > 0:
        pred_points_img = ProjectWorldToImage(pred_waypoints_3d_valid.tolist(), cam_to_ego, ego_to_world)
        
        if len(pred_waypoints_3d_valid) > 1:
            pred_left_3d = OffsetTrajectory3D(pred_waypoints_3d_valid, -1.73 / 2)
            pred_right_3d = OffsetTrajectory3D(pred_waypoints_3d_valid, 1.73 / 2)
            pred_left_img = ProjectWorldToImage(pred_left_3d.tolist(), cam_to_ego, ego_to_world)
            pred_right_img = ProjectWorldToImage(pred_right_3d.tolist(), cam_to_ego, ego_to_world)
            
            pred_polygon = np.vstack((np.array(pred_left_img), np.array(pred_right_img)[::-1])).astype(np.int32)
            frame_pred = np.zeros_like(img)
            cv2.fillPoly(frame_pred, [pred_polygon], color=(0, 125, 255))
            mask_pred = frame_pred.astype(bool)
            img[mask_pred] = cv2.addWeighted(img, 0.5, frame_pred, 0.5, 0)[mask_pred]
        
        # Draw predicted waypoints
        for pt in pred_points_img:
            cv2.circle(img, tuple(pt.astype(int)), radius=8, color=(0, 125, 255), thickness=-1)
    
    # Add legend
    cv2.putText(img, 'Green: Ground Truth', (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    cv2.putText(img, 'Orange: Predicted', (10, 70), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 125, 255), 2)
    
    # Display
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(16, 9))
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.title(f'Sample {sample_idx}: Trajectory Prediction')
    plt.show()
    
except Exception as e:
    print(f"Visualization failed: {e}")
```

## 8. Run Full Evaluation on N Samples

Evaluate multiple samples and compute aggregate metrics.

```python
num_samples = 50  # Adjust as needed

results = []

for idx in tqdm(range(num_samples), desc='Evaluating'):
    image, ego_positions, gt_waypoints, cam_to_ego, ego_to_world = get_sample_data(idx)
    pixel_values = model.image_processor(images=[image], return_tensors='pt').pixel_values.to(device)
    
    with torch.no_grad():
        outputs, gen_texts = model.generate_trajectory(pixel_values, [ego_positions])
    
    gen_text = gen_texts[0]
    pred_coords = parse_coords_from_text(gen_text, max_points=10)
    
    num_valid_waypoints = pred_coords.shape[0]
    format_compliant = (num_valid_waypoints == 10)
    
    if pred_coords.shape[0] < 10:
        pad = np.full((10 - pred_coords.shape[0], 2), np.nan)
        pred_coords = np.vstack([pred_coords, pad])
    elif pred_coords.shape[0] > 10:
        pred_coords = pred_coords[:10]
    
    diffs = pred_coords - gt_waypoints
    l2_per_waypoint = np.linalg.norm(diffs, axis=1)
    ade = np.nanmean(l2_per_waypoint)
    fde = l2_per_waypoint[-1]
    miss_rate_10m = 1.0 if fde > 10.0 else 0.0
    
    results.append({
        'idx': idx,
        'num_valid_waypoints': num_valid_waypoints,
        'format_compliant': format_compliant,
        'ade': ade,
        'fde': fde,
        'miss_rate_10m': miss_rate_10m
    })

print(colored("✓ Evaluation complete!", "green"))
```

## 9. Display Aggregate Results

```python
# Aggregate metrics
ades = [r['ade'] for r in results if not np.isnan(r['ade'])]
fdes = [r['fde'] for r in results if not np.isnan(r['fde'])]
miss_rates = [r['miss_rate_10m'] for r in results]
format_compliance = [r['format_compliant'] for r in results]

print(colored("=" * 50, "cyan"))
print(colored("EVALUATION SUMMARY", "cyan", attrs=['bold']))
print(colored("=" * 50, "cyan"))
print(f"Samples evaluated: {len(results)}")
print(f"\nTrajectory Metrics:")
print(f"  ADE (mean): {np.mean(ades):.4f} ± {np.std(ades):.4f} m")
print(f"  FDE (mean): {np.mean(fdes):.4f} ± {np.std(fdes):.4f} m")
print(f"  Miss Rate @ 10m: {np.mean(miss_rates):.2%}")
print(f"  Format Compliance: {np.mean(format_compliance):.2%}")
print(colored("=" * 50, "cyan"))
```

## 10. Quick Checkpoint Switching

To test a different checkpoint **without reloading the model**, just run this cell:

```python
# Switch to a different checkpoint (FAST - model stays in GPU)
new_checkpoint_path = "./experiments/runs/qwenclip1a/checkpoint_latest.pth"

data = torch.load(new_checkpoint_path, map_location=device)
if 'language_model_state_dict' in data:
    model.vision_tower.load_state_dict(data['vision_tower_state_dict'])
    model.mlp_projector.load_state_dict(data['model_state_dict'])
    print(colored(f"✓ Switched to: {new_checkpoint_path}", "green"))
else:
    model.mlp_projector.load_state_dict(data)
    print(colored(f"✓ Switched to: {new_checkpoint_path}", "green"))

# Now go back and re-run cells 6-9 to evaluate the new checkpoint!
```
